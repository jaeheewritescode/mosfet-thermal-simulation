# MOSFET Loss Equations and Assumptions

## Loss equations

### Conduction loss

`P_cond = I_RMS² × R_DS(on)`

Two on-resistance values are considered:

- **0.0139 ohm**: LTspice-model baseline used for direct comparison with LTspice
- **0.0175 ohm**: maximum datasheet value at 25°C used for the conservative case

Calculated conduction losses:

- **Model-based case: P_cond = 0.695 W**
- **Datasheet maximum case: P_cond = 0.875 W**

### Switching loss

`P_sw = 0.5 × V_DS × I_D × (t_r + t_f) × f_s`

Calculated switching loss:

**P_sw = 0.630 W**

The switching loss is the same in both cases because this equation does not depend on R_DS(on).

### Total analytical loss

`P_total = P_cond + P_sw`

- **Model-based analytical total: 1.325 W**
- **Datasheet maximum analytical total: 1.505 W**

## LTspice result

The average MOSFET power calculated from the exported voltage and current waveforms was:

**P_LTspice = 0.731 W**

The direct analytical-to-LTspice comparison uses the **0.0139 ohm model-based case**. The analytical result of **1.325 W** is approximately **44.8% higher** than the LTspice result.

## Main assumptions

- Input voltage: **24 V**
- Output voltage: **12 V**
- Output current: **10 A**
- Switching frequency: **50 kHz**
- Duty cycle: **0.5**
- MOSFET: **IRFZ44N**
- Gate-drive voltage: **10 V**
- Model-based on-resistance: **0.0139 ohm**
- Datasheet maximum on-resistance at 25°C: **0.0175 ohm**
- Rise time: **60 ns**
- Fall time: **45 ns**
- LTspice simulation temperature: **25°C**

The analytical switching-loss equation assumes linear overlap between MOSFET voltage and current during switching. The datasheet switching times may not exactly match the LTspice operating conditions.

The LTspice gate source is ideal, so the simulated switching transitions may be faster than those of a real gate driver. Gate-driver loss, PCB resistance and stray inductance are not included.

## Thermal-model input

The **1.505 W datasheet maximum case** will be used as the conservative baseline heat input.

The **1.325 W model-based analytical result** and the **0.731 W LTspice result** will be used as comparison cases.
